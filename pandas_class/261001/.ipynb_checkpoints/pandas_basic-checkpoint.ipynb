```python
doc.describe() # 전체데이터를 중요한 지표만 추출해서 묘사 -> count
# 5 number summary
# FIPS = Federal Information Processing Standards Code = 미국 연방 지역 식별코드
# FIPS => 36061 : 뉴욕시 맨해튼 // 06037 => 로스앤젤레스
# 데이터를 수집 -> 분석 -> 높은 확률 -> 상관관계 따져볼 수 있음
# 현재 우리의 데이터가 상관관계를 따져볼 수 있는 포맷(타입) 구성? => 수학공식 => 수치화된 값
```

```python
doc.corr(numeric_only=True) # 문자열이 아닌 수치화된 데이터만 가져와서 상관관계!!!
```
