##### 5) 서로 다른 데이터프레임간 병합
> 국가별 코로나 확진자수 추이 조회 : 2020-01-01 ~ 2023-12-31 : 하나의 데이터 프레임안에 각 개별 컬럼으로 존재
> 각 날짜의 데이터는 별도의 독립적인 csv
> 각 국가별 이름, 확진, 사망, 회복, 국가별 축약키워드 : 하나의 데이터 프레임안에 모든 데이터들이 개별 컬럼

```python
import pandas as pd

df1 = pd.DataFrame({
    "id": [1, 2, 3],
    "customer_id": [1, 2, 3],
    "customer_name": ["Robert", "Peter", "Dave"]
})

df1
```

```python
df2 = pd.DataFrame({
    "id": [1, 2, 4],
    "order_id": [100, 200, 300],
    "order_date": ["2021-01-21", "2021-02-03", "2020-10-01"]
})

df2
```

```python
# 서로 다른 데이터 프레임 2개를 생성 > 한개의 프레임으로 병합?!
# 병합의 방식 2가지
# 상.하
# 좌.우
```

```python
pd.concat([df1, df2], axis=0)
# 서로 다른 데이터프레임을 상.하 기준으로 -> 동일한 컬럼명을 가지고 있는 경우에만 정상데이터, NaN
# axis = 0 디폴트 값 => 상하 병합 axis = 1 => 좌우 병합
```

```python
pd.concat([df1, df2], axis=1) # 물리적으로 교집합이 없더라도 일단 좌.우 병합
```

```python
pd.merge(df1, df2) # 공통요소 = 컬럼을 기준으로 좌.우 병합 = INNER JOIN # on을 작성하지 않았어도 알아서 공통 컬럼명
```

```python
pd.merge(df1, df2, on="id") # SQL, JOIN 특정 테이블간 공통 컬럼 내 ON 전치사를 활용
```

```python
pd.merge(df1, df2, on="id", how="inner")
```

```python
pd.merge(df1, df2, on="id", how="outer") # left right 모두 살린 경우
```

```python
pd.merge(df1, df2, on="id", how="left") # left
```

```python
pd.merge(df1, df2, on="id", how="right") # right
```

```python
# 지금까지는 서로 다른 데이터프레임을 하나로 병합하는 조건, 특정 컬럼을 기준!!!
# 인덱스를 기준으로 병합

df1 = df1.set_index("id")
df1
```

```python
df2 = df2.set_index("id")
df2
```

```python
pd.merge(df1, df2, left_index=True, right_index=True)
```

```python
pd.merge(df1, df2, how="outer", left_index=True, right_index=True)
# 인덱스를 기준으로 병합을 하되, 레프트, 라이트 모두 아우터 생존
```

```python
pd.merge(df1, df2, how="left", left_index=True, right_index=True)
# 인덱스를 기준으로 병합을 하되, 레프트 아우터 생존
```

```python
pd.merge(df1, df2, how="right", left_index=True, right_index=True)
# 인덱스를 기준으로 병합을 하되, 라이트 아우터 생존
```
